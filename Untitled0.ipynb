```python
# Student Performance Analysis using CSV File in Google Colab

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('/content/Student_Performance_Dataset.csv')

df
```

```python
# Display first 5 rows
print("First 5 Rows:")
df.head()

```

```python
# Dataset information
print("\nDataset Info:")
df.info()

```

```python
# Basic statistics
print("\nStatistical Summary:")
df.describe()

```

```python


# Highest Percentage marks
highest_percentage = df['Final_Percentage'].max()
print("Highest percentage :", highest_percentage)

# Lowest lowest percentage
lowest_percentage = df['Final_Percentage'].min()
print("Lowest percentage:", lowest_percentage)
```

```python
print(df.corr(numeric_only=True))
```

```python
corr = df.corr(numeric_only=True)

final_corr = corr['Final_Percentage'].drop('Final_Percentage')

plt.figure(figsize=(10,6))

plt.bar(
    final_corr.index,
    final_corr.values,
    color='skyblue'
)

plt.xticks(rotation=45)

plt.title("Correlation with Final Percentage")
plt.xlabel("Features")
plt.ylabel("Correlation")

plt.show()
```

```python
df['Performance_Level'].value_counts().plot(
    kind='bar',
    color='skyblue'
)

plt.title("Performance Level Distribution")
plt.xticks(rotation=45)
plt.xlabel("Performance Level")
plt.ylabel("Count")

plt.show()
```

```python
df['Pass_Fail'].value_counts().plot(
    kind='pie',
    color=['green', 'red']
)

plt.title("Pass vs Fail")


plt.show()
```

```python
plt.scatter(
    df['Study_Hours_Per_Day'],
    df['Final_Percentage'],
    color='orange'
)

plt.title("Study Hours vs Final Percentage")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Final Percentage")

plt.show()
```

```python
high_performer=df[df['Final_Percentage']>=df['Previous_Year_Score']]
high_performer= high_performer[['Student_ID','Final_Percentage']]
high_performer
```
